```python
import pandas as pd

df = pd.read_csv("AI_Usage_and_Impact_on_Students_and_Professionals.csv")

hours_cols = ["AI_Usage_Hours_Per_Day", "Work_or_Study_Hours_Per_Day"]

def clean(df):
    df = df.drop_duplicates().copy()

    # Text columns: trim spaces, consistent case
    df["Gender"] = df["Gender"].str.strip().str.title()
    df["AI_Tool"] = df["AI_Tool"].str.strip().replace({"Chat GPT": "ChatGPT"})

    # Numeric columns: pull the number out of strings like "2 hrs"
    for col in hours_cols:
        df[col] = pd.to_numeric(df[col].astype(str).str.extract(r"([\d.]+)")[0])

    return df

df_clean = clean(df)
```

```python
df_clean["Age"] = df_clean["Age"].str.replace(" yrs", "")
df_clean["Monthly_Income"] = df_clean["Monthly_Income"].str.replace("$", "").str.replace(",", "").str.strip()
df_clean["Monthly_AI_Cost"] = df_clean["Monthly_AI_Cost"].str.replace("$", "").str.strip()
df_clean["Time_Saved_Hours_Per_Week"] = df_clean["Time_Saved_Hours_Per_Week"].str.replace(" hrs", "")
```

```python
# Text columns: fill empty values with "Unknown"
text_cols = ["Gender", "Country", "Education_Level", "Profession",
             "AI_Tool", "AI_Purpose", "Tasks_Performed", "Would_Recommend"]
for col in text_cols:
    df_clean[col] = df_clean[col].fillna("Unknown")

# Number columns: turn into numbers, then fill empty values with the median
num_cols = ["Age", "Monthly_Income", "AI_Usage_Hours_Per_Day", "Monthly_AI_Cost",
            "Productivity_Score", "Accuracy_Rating", "Satisfaction_Score",
            "Time_Saved_Hours_Per_Week", "Work_or_Study_Hours_Per_Day"]
for col in num_cols:
    df_clean[col] = pd.to_numeric(df_clean[col], errors="coerce")
    df_clean[col] = df_clean[col].fillna(df_clean[col].median())
```
